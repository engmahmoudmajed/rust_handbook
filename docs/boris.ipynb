````rust
// continue keyword -> force loop to move the next iteration
fn main() {

  let mut seconds = 10 ;

  loop{
      if seconds <= 0 {
      println!("Blastoff...");
      break;
    }
      if seconds % 2 == 0 {
      println!("{seconds} seconds (even number),skipping 3 second...");
      seconds -= 3;
      continue;
    }
    println!("{seconds} seconds to blass off....");
    seconds -= 1;
  }

}
````

````rust


// while loop continue iterating as long condtion is met
fn main() {
  let mut seconds = 10 ;
  while seconds > 0 {
      if seconds % 2 == 0 {
      println!("{seconds} seconds (even number),skipping 3 second...");
      seconds -= 3;
      continue;
    }
    println!("{seconds} seconds to blass off....");
    seconds -= 1;
  }
  println!("Blastoff...");
}
````

##### Recursion

````rust
fn countdown(n: i32) {
    // Base case: stop when n is 0 or less
    if n <= 0 {
        println!("Liftoff!");
        return;
    }
    // Print the current number
    println!("{}", n);
    // Recursive step: call function with n - 1
    countdown(n - 1);
}

fn main() {
    countdown(3);
}
````

````rust
fn main() {
	//Move is transfare ownership from owner to another one 
	let person = String::from("Mahmoud");
	// what happen here 
	// not impelement copy trait here what happen is diffrent 
	// why its just copy -> because will aduplicate and consume more memorey
	// what happen here its copy the stack entery (its copy refrence to heap data and length and capcity)
	// back to ownershop
	// now is there are one piece of data on heap who is responsible for cleaning it up fo deallocating
	// when we assign genius =  person // we move ownership from person to  genius 
	// but this this mean after next line we can't use person because its out of the scope 
	// what mean one owner for value 
	// in other language 

/*  	
	For example, in other programming languages, without the concept of ownership, there could be two
	references pointing to the same heap data with this kind of code.
	So then when the function scope would end, the program would remove the 'person' variable, it would
	become out of scope, and it would try to remove the contents on the heap.
	But then that program would also have the 'genius' variable going out of scope, and that would attempt
	to clear the location and memory that was already cleared by the previous variable going out of scope.
	So when the language tries to free available memory on the heap twice, it actually runs into an error
	because that heap allocation will technically no longer have a value.
	This is called a double free error
	in other programming languages, and it can lead to memory corruption and security vulnerabilities.
	But it's not possible
	within Rust. We can't have two variables, two owners, attempting to clean up the same heap data because
	we can't have two owners.
*/
	println!("my name is {person}"); 

	let genius = person; //

	println!("");
	// println!("my name is {person}"); // this wil make error borrow of moved value 


}

````

| Aspect | `&T` (immutable reference) | `&mut T` (mutable reference) |
|---|---|---|
| What it is | A borrowed pointer to data you don't own | A borrowed pointer to data you don't own, with write permission |
| Read access | Yes | Yes |
| Write access | No | Yes |
| How many can exist at once | Multiple, simultaneously | Only one, and no `&T` allowed at the same time |
| Ownership | Stays with original owner | Stays with original owner |
| Underlying rule | "Shared read access" — safe because nothing can change while others read | "Exclusive access" — prevents data races since only one writer exists |
| Syntax | `fn f(x: &String)` | `fn f(x: &mut String)` |
| Call site | `f(&s)` | `f(&mut s)` |
| Best use case | You only need to inspect/use a value — printing, comparing, reading fields, passing to something that just needs to look at the data | You need to modify the value in place — appending, clearing, sorting, updating fields — without taking or giving back ownership |
| Typical example | `fn print_meal(meal: &String)` | `fn add_flour(meal: &mut String)` |
| Performance reason to prefer it | Avoids copying large data just to read it | Avoids the cost of taking full ownership + returning it back at the end |

**Rule of thumb for choosing:**
- Default to `&T` first — it's the least restrictive and safest option.
- Only use `&mut T` when the function genuinely needs to change the caller's original data.
- Never take ownership (`T`) just to read or mutate temporarily — that's needless, since you'd have to hand ownership back (usually via return) to keep using the value afterward.

````rust
// Define Methods
// Method : function that belongs to an instance
// like .push .pull
// method like function -> accept params return values
// but like we add behavior owend by struct

// use debug to print struct
#[derive(Debug)]

struct Song {
    title: String,
    relase_year: u32,
    duration: u32, // duration in seconds
}

// we use impl to implement methods for struct ( add methods for struct)
impl Song {
    // Immuutable struct value (self parameters takes ownership)
    /*
    display_song_info(self:Song)
    display_song_info(self:Self) //best for all
    */
    // Mutable struct value (self mut parameters takes ownership,has permission to change the value of struct)
    /*
    display_song_info(mut self)
    display_song_info(self:mut Song)
    display_song_info(self:mut Self) //best for all
    */
    // immutable reference to struct value (&self parameters borrows the value of struct)
    /*
    display_song_info(&self)
    display_song_info(self:&Song)
    display_song_info(self:&Self) //best for all
    */
    // mutable reference to struct value (&mut self parameters borrows the value of struct, has permission to change the value of struct)
    /*
    display_song_info(&mut self)
    display_song_info(self:&mut Song)
    display_song_info(self:&mut Self) //best for all
    */
    fn display_song_info(&self) {
        println!(
            "Song: {}, Year: {}, Duration: {} seconds",
            self.title, self.relase_year, self.duration
        );
    }
}

fn main() {
    println!("------------ Run Rust App ---------------");
    let song_1 = Song {
        title: "Bohemian Rhapsody".into(),
        relase_year: 1975,
        duration: 345,
    };
    song_1.display_song_info();
}
````

````rust
// Design pattern is a recommended way to write or struct code to solve a specific problemS
// Design pattern is a recommended way to write or struct code to solve a specific problem in a specific context. It is a reusable solution to a common problem that can be applied in different situations. Design patterns are not code snippets or libraries, but rather they are general concepts and best practices that can be implemented in various programming languages. They provide a way to structure code in a way that is maintainable, scalable, and easy to understand. Some common design patterns include the Singleton pattern, the Observer pattern, the Factory pattern, and the Strategy pattern.
//
/*
*  in buider pattern
*  we have every method return the instance of itself,or
*  or alternatively we can return arefrence to the instance of itself,
*  Advantege of this design pattern
* it allow us to chain the methods in sequince in a very elegant manar
* it allow us to chain the methods in together in a fluent interface, making the code more readable and expressive.
*
*
*
*/

#[derive(Debug)]

struct Computer {
    cpu: String,
    memory: u32,
    hard_drive_capacity: u32,
}

impl Computer {
    fn new(cpu: String, memory: u32, hard_drive_capacity: u32) -> Self {
        Computer {
            cpu,
            memory,
            hard_drive_capacity,
        }
    }
    //---
    fn upgrade_cpu(&mut self, new_cpu: String) {
        self.cpu = new_cpu;
    }
    fn upgrade_memory(&mut self, new_memory: u32) {
        self.memory = new_memory;
    }
    fn upgrade_hard_drive(&mut self, new_hard_drive_capacity: u32) {
        self.hard_drive_capacity = new_hard_drive_capacity;
    }
}

fn main() {
    let mut computer = Computer::new("Intel i7".into(), 16, 1000);
    println!("{:#?}", computer);
    computer.upgrade_cpu("Intel i9".into());
    println!("{:#?}", computer);
    computer.upgrade_memory(32);
    println!("{:#?}", computer);
}


//------------ time for a new design pattern called the builder pattern
//
/*
*  in buider pattern
*  we have every method return the instance of itself,or
*  or alternatively we can return arefrence to the instance of itself,
*  Advantege of this design pattern
* it allow us to chain the methods in sequince in a very elegant manar
* it allow us to chain the methods in together in a fluent interface, making the code more readable and expressive.
*
*
*
*/

#[derive(Debug)]

struct Computer {
    cpu: String,
    memory: u32,
    hard_drive_capacity: u32,
}

impl Computer {
    fn new(cpu: String, memory: u32, hard_drive_capacity: u32) -> Self {
        Computer {
            cpu,
            memory,
            hard_drive_capacity,
        }
    }
    //---
    fn upgrade_cpu(&mut self, new_cpu: String) ->&mut Self {
        self.cpu = new_cpu;
        self
    }
    fn upgrade_memory(&mut self, new_memory: u32) ->&mut Self {
        self.memory = new_memory;
        self
    }
    fn upgrade_hard_drive(&mut self, new_hard_drive_capacity: u32) ->&mut Self {
        self.hard_drive_capacity = new_hard_drive_capacity;
        self
    }
}

fn main() {
    let mut computer = Computer::new("Intel i7".into(), 16, 1000);
    println!("{:#?}", computer);
    computer
    .upgrade_cpu("Intel i9".into())
    .upgrade_memory(32)
    .upgrade_hard_drive(2000);

    println!("{:#?}", computer);
}




````

````rust
#[derive(Debug)]

enum PaymentMethodType {
    CreditCard(String),
    DepitCard(String),
    PayPal(String, String),
}

// like paypal will need password and email

fn main() {
    let visa = PaymentMethodType::CreditCard(String::from("6666-6666-6666-6666"));
    let mut payment_method =
        PaymentMethodType::PayPal(String::from("maj@gmail.com"), String::from("pass"));
    println!("{:?}", payment_method)
}

//--------------
/*
 * An enum can have multiple variants, each holding different types and amounts of associated data, which in turn require different amounts of memory.
	Rust handles this by allocating memory for the enum based on its largest variant (the worst-case scenario).
	Example: a String on the stack takes up 24 bytes (since it's really just a pointer to the actual text data stored on the heap).
	If a variant like CreditCard or DebitCard holds one String, it needs at least 24 bytes.
	If a variant like PayPal holds two Strings, it needs 48 bytes (24 × 2).
	Rust picks the largest size among all variants (here, 48) so that any variant can be stored without needing to reallocate memory if the variable is later reassigned to a different variant.
	The enum's total size isn't just the sum of all variants' data — Rust also needs to store the variant tag (like which one it is: PayPal, CreditCard, etc.), and it sometimes applies additional optimizations.
	So the enum won't necessarily be exactly 48 bytes, but it will be at least 48 bytes.
	Day to day, you don't need to worry about stack vs. heap, exact memory sizes, or byte-level optimization — computers are fast enough today. It's just good to know that Rust handles all of this under the hood to keep your program efficient.
 *
*/
````

##### simplified summary of the lesson on generic `impl` blocks in Rust:


###### The Core Idea

In the previous lesson, you learned how to define methods for a struct with a specific concrete type (like `String` or an array). Today's goal is to define methods that work **regardless of what type T is**.

###### The Problem

Given:
```rust
struct TreasureChest<T> {
    captain: String,
    // ...
}
```

You can't write:
```rust
impl TreasureChest {}      // ❌ wrong — missing the type
impl TreasureChest<T> {}   // ❌ also wrong
```

Because in the second case, Rust assumes `T` refers to an **actual concrete type** that already exists (as if there were a real struct or enum named `T`), not a generic placeholder.

###### The Solution — Correct Syntax

```rust
impl<T> TreasureChest<T> {
    fn capital_captain(&self) -> String {
        self.captain.to_uppercase()
    }
}
```

By adding `<T>` right after the `impl` keyword, you're telling Rust:
> "I'm declaring a generic called T, and I'm going to use it below."

This makes Rust understand that `T` is a **generic placeholder**, not a specific named type.

###### The Result

This method (`capital_captain`) will now work on **any instance** of `TreasureChest<T>`, no matter what type fills in T:

| Variable | Type of T | Result |
|---|---|---|
| `gold_chest` | `&str` | "FIREBEARD" |
| `silver_chest` | `String` | "BLOODSAIL" |
| `special_chest` | another type | "BOOTYPLUNDER" |

###### Key Takeaway
- `impl<T> Struct<T>` → generic methods that work for **all** types
- `impl Struct<String>` → methods that only work when T is specifically `String`

##### Generic in Enums

##### Here's a simplified summary of the lesson on **generics with enums** in Rust:

###### The Core Idea

Just like structs, enums can also use generics. This lesson shows an example enum `Cheesesteak<T>` where the generic type `T` isn't required in every variant — you decide where it's used.

```rust
enum Cheesesteak<T> {
    Plain,
    Topping(T),
}
```

- `Plain` → no associated data, doesn't touch `T` at all
- `Topping(T)` → holds one piece of data, whose type is generic `T`

###### Type Inference Works When T Is Used

When you create a `Topping` variant, Rust can **infer** the type of `T` from the value you give it:

```rust
let mushroom = Cheesesteak::Topping("mushroom");        // T = &str
let onions   = Cheesesteak::Topping("onion".to_string()); // T = String
let bacon    = Cheesesteak::Topping(&topping);            // T = &String
```

Each of these ends up with a **different concrete type** for `T`, depending on what was passed in.

###### The Problem With `Plain`

```rust
let plain = Cheesesteak::Plain; // ❌ won't compile
```

Even though `Plain` doesn't carry any data, Rust still needs to know what `T` *would* be — because:
- The variable might later be reassigned to a `Topping` variant
- The compiler must know `T`'s type in advance to check future assignments are valid

Rust's error message will literally ask you to specify the type.

###### The Fix — Explicit Type Annotation

```rust
let plain: Cheesesteak<String> = Cheesesteak::Plain; // ✅ works
```

Now Rust knows: "if this variable ever becomes a `Topping`, the data inside **must** be a `String`."

###### Enforcement Example

```rust
let mut plain: Cheesesteak<String> = Cheesesteak::Plain;
plain = Cheesesteak::Topping("bacon"); // ❌ error: expected String, found &str
plain = Cheesesteak::Topping("bacon".to_string()); // ✅ fixed
```

Since `T` was locked in as `String`, Rust rejects anything that isn't a `String` — keeping your code type-safe.

###### Key Takeaways
| Situation | Does Rust need annotation? |
|---|---|
| Variant that uses `T` (e.g. `Topping(value)`) | ❌ No — inferred from the value |
| Variant that doesn't use `T` (e.g. `Plain`) | ✅ Yes — must be explicit |

This is because Rust needs to know `T`'s type from the start, both for memory purposes and to validate any future changes to the variable.

````rust

// Generic with enum 
#[derive(Debug)]
enum Cheesesteak<T>{
	Plain,
	Topping(T)
}

fn main() {
	let mushroom = Cheesesteak::Topping("mushroom");
	let onions =  Cheesesteak::Topping("onion".to_string());
	// next code will not compile 
	// type annotations needed for `Cheesesteak<_>`rustcClick for full compiler diagnostic
	// let plain:Cheesesteak<String> = Cheesesteak::Plain;
	println!("{:#?}",mushroom);
	println!("{:#?}",onions);
	// println!("{:#?}",plain);
	let mut plain:Cheesesteak<String> = Cheesesteak::Plain;
	plain = Cheesesteak::Topping(String::from("sausage"));
	println!("{:#?}",plain);
}
````
