##### loops

```rust
// loop and break keyword

fn main() {
  let mut seconds = 10 ;
  loop{
      if seconds == 0 {
      println!("Blastoff...");
      break;
    }
    println!("{seconds} seconds to blass off....");
    seconds -= 1;
  }
}
```

```rust
// continue keyword -> force loop to move the next iteration
fn main() {

  let mut seconds = 10 ;

  loop{
      if seconds <= 0 {
      println!("Blastoff...");
      break;
    }
      if seconds % 2 == 0 {
      println!("{seconds} seconds (even number),skipping 3 second...");
      seconds -= 3;
      continue;
    }
    println!("{seconds} seconds to blass off....");
    seconds -= 1;
  }

}
```

```rust


// while loop continue iterating as long condtion is met
fn main() {
  let mut seconds = 10 ;
  while seconds > 0 {
      if seconds % 2 == 0 {
      println!("{seconds} seconds (even number),skipping 3 second...");
      seconds -= 3;
      continue;
    }
    println!("{seconds} seconds to blass off....");
    seconds -= 1;
  }
  println!("Blastoff...");
}
```

```rust

fn countdown(seconds:i32){

  if seconds == 0 {
    println!("blass off....");
  }else{
    println!("{seconds} seconds to blass off....");
    countdown(seconds - 1);
  }
}


// Recursion -> function call it self and stop/work by condtion 
// the base case is the condtion stop recursion
fn main() {
  countdown(5);

}
```
